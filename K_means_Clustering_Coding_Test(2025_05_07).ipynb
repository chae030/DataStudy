```python
#Q2 : Enter your answer in this cell.

import matplotlib.pyplot as plt
from sklearn.cluster import KMeans  # K-means 클러스터링 알고리즘

# 이미 제공된 데이터 (신장과 체중 정보가 담긴 2차원 배열 df_hw 사용)

sse = []  # SSE (Sum of Squared Errors, 오차 제곱합)를 저장할 리스트
k_range = range(1, 11)  # K (군집 개수)를 1부터 10까지 순차적으로 시도

# Elbow Method를 위한 반복문
for k in k_range:
    # KMeans 모델 생성
    # n_clusters=k : 클러스터 개수를 k로 설정
    # random_state=0 : 랜덤 초기화 시 항상 같은 결과를 얻기 위해 사용 (재현성 확보)
    #  → 머신러닝 알고리즘은 랜덤하게 초기값을 설정하기 때문에, 
    #     random_state를 고정하면 결과가 항상 동일함.
    kmeans = KMeans(n_clusters=k, random_state=0)
    
    # KMeans 모델에 데이터 학습 (클러스터링 수행)
    kmeans.fit(df_hw)
    
    # inertia_ 속성: 현재 모델의 SSE (오차 제곱합) 값
    sse.append(kmeans.inertia_)  # 계산된 SSE 값을 리스트에 추가

# 📊 Elbow Method 결과 시각화
plt.plot(k_range, sse, marker='o')  # x축: K값, y축: SSE 값, 동그라미로 점 찍음
plt.xlabel('Number of Clusters (K)')  # x축 라벨
plt.ylabel('SSE (Inertia)')  # y축 라벨 (SSE 값)
plt.title('Elbow Method for Optimal K')  # 그래프 제목
plt.grid(True)  # 격자 표시 (그래프 읽기 쉽게)
plt.show()  # 그래프 출력
```

*Q3.* **Use "Silhouette Coefficient" to find the "K"(target data = df_hw)**

```python
#Q3 : Enter your answer in this cell.

from sklearn.metrics import silhouette_score  # 실루엣 계수 계산 함수

s_scores = []  # 실루엣 계수 저장 리스트
k_range = range(2, 11)  # 실루엣 계수는 K=2 이상부터 계산 가능 (1개 군집은 무의미)

for k in k_range:
    # KMeans 모델 생성 및 학습
    kmeans = KMeans(n_clusters=k, random_state=0)
    kmeans.fit(df_hw)
    
    # 군집 레이블 예측
    labels = kmeans.labels_
    
    # 실루엣 계수 계산
    # df_hw: 데이터, labels: 각 데이터가 속한 군집 번호
    score = silhouette_score(df_hw, labels)
    s_scores.append(score)  # 실루엣 계수 저장

# 📊 결과 시각화
plt.plot(k_range, s_scores, marker='o')
plt.xlabel('Number of Clusters (K)')  # x축 라벨
plt.ylabel('Silhouette Coefficient')  # y축 라벨 (실루엣 계수 값)
plt.title('Silhouette Method for Optimal K')
plt.grid(True)
plt.show()

# 📝 최적 K 출력
optimal_k = k_range[s_scores.index(max(s_scores))]
print(f"실루엣 계수가 가장 높은 최적 K값은: {optimal_k}")
```

*Q4.* **Plot the scatter points and cluster centroids(*Both*) using the matplotlib library.**

```python
#Q4 : Enter your answer in this cell.

s_scores = []
k_range = range(2, 11)

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=0)
    kmeans.fit(df_hw)
    labels = kmeans.labels_
    score = silhouette_score(df_hw, labels)
    s_scores.append(score)

optimal_k = k_range[s_scores.index(max(s_scores))]
print(f"최적 K 값: {optimal_k}")

# ✅ 2. 최적 K값으로 KMeans 클러스터링 수행
kmeans_final = KMeans(n_clusters=optimal_k, random_state=0)
kmeans_final.fit(df_hw)
labels_final = kmeans_final.labels_
centroids = kmeans_final.cluster_centers_

# ✅ 3. 시각화 (산점도 + 중심점 표시)
plt.figure(figsize=(8, 6))

# 군집에 따라 색상 다르게 표시
for i in range(optimal_k):
    cluster_points = df_hw[labels_final == i]
    plt.scatter(cluster_points[:, 0], cluster_points[:, 1], label=f'Cluster {i+1}', s=30)

# 군집 중심점 (Centroids) 표시
plt.scatter(centroids[:, 0], centroids[:, 1], 
            marker='*', color='red', s=300, label='Centroids')

plt.xlabel('Height')
plt.ylabel('Weight')
plt.title('Cluster Visualization with Centroids')
plt.legend()
plt.grid(True)
plt.show()
```
