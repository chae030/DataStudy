<h3>K-means Clustering

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# 9개 도시 좌표 (왼쪽 그림 기준, 0부터 시작하는 좌표계 사용)
city_coords = np.array([
    [0, 0],
    [0, 2],
    [0, 3],
    [1, 6],
    [2, 1],
    [2, 6],
    [3, 7],
    [6, 7],
    [8, 5]
])

# 초기 중심점 설정 (오른쪽 그림 기준, 별표 위치)
# 삼각형, 사각형, 다이아몬드 순서로 
initial_centroids = np.array([
    [2, 7],  # 삼각형
    [1, 3],  # 사각형
    [8, 3]   # 다이아몬드
])

# KMeans 모델 생성 및 학습 (초기 중심 고정, 1회 초기화)
kmeans = KMeans(n_clusters=3, init=initial_centroids, n_init=1, random_state=0)
kmeans.fit(city_coords)

# 각 도시가 속한 클러스터
labels = kmeans.labels_

# 시각화 준비
markers = ['^', 's', 'D']  # 삼각형, 사각형, 다이아몬드
colors = ['orange', 'magenta', 'navy']

plt.figure(figsize=(12, 6))

# 원본 도시 위치 (왼쪽 그림)
plt.subplot(1, 2, 1)
plt.scatter(city_coords[:, 0], city_coords[:, 1], color='black', s=100)
plt.grid(True)
plt.xticks(range(0, 10))
plt.yticks(range(0, 10))
plt.title("도시 원래 위치")

# 클러스터링 결과 (오른쪽 그림)
plt.subplot(1, 2, 2)
for i in range(3):
    cluster_points = city_coords[labels == i]
    plt.scatter(cluster_points[:, 0], cluster_points[:, 1],
                marker=markers[i], color=colors[i], s=200, label=f'클러스터 {i+1}')
    
# 초기 중심점 별표 표시
plt.scatter(initial_centroids[:, 0], initial_centroids[:, 1], marker='*', s=300)

plt.grid(True)
plt.xticks(range(0, 10))
plt.yticks(range(0, 10))
plt.show()
```
